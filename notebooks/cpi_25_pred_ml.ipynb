# Prediction for the same dataset of year 2025 using the datset of those of year 2021 to 2024

```python
# ============================================================
# CPI 2024 PREDICTION USING HISTORICAL TREND
# Training: 2014-2019 + 2022-2023
# COVID years 2020-2021 excluded
# Test/Prediction year: 2024
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

```

```python
df1= pd.read_csv(r"C:\Users\himan\Education1\Projects\MoSPI_CPI\Data\merged\merged_cpi_rebased_2012.csv")
df1.head()
```

```python
print("\nColumns:")
print(df1.columns.tolist())
```

```python
# ============================================================
# 4. MAKE SURE YEAR AND CPI ARE NUMERIC
# ============================================================

df1["Year_i"] = pd.to_numeric(
    df1["Year_i"],
    errors="coerce"
)

df1["Index_i"] = pd.to_numeric(
    df1["Index_i"],
    errors="coerce"
)

# Remove rows where year or CPI is missing
df1 = df1.dropna(
    subset=["Year_i", "Index_i"]
)

df1["Year_i"] = df1["Year_i"].astype(int)
```

```python
# ============================================================
# 5. CLEAN STRING COLUMNS
# ============================================================

string_columns = [
    "Month_i",
    "State_i",
    "Sector",
    "Group_i",
    "SubGroup"
]

for col in string_columns:
    df1[col] = df1[col].astype(str).str.strip()
```

```python
# ============================================================
# 6. DEFINE THE CPI SERIES
# ============================================================

# Each unique combination represents one CPI series.

keys = [
    "Month_i",
    "State_i",
    "Sector",
    "Group_i",
    "SubGroup"
]

print("\nNumber of unique CPI series:")
print(df1[keys].drop_duplicates().shape[0])
```

```python
# ============================================================
# 7. DEFINE TRAINING AND TEST YEARS
# ============================================================
# 2020 and 2021 = COVID period
#
# Training:
# 2014-2019 and 2022-23
# Test:
# 2024

training_years = [
    2014,
    2015,
    2016,
    2017,
    2018,
    2019,
    2022,
    2023
]

test_year = 2024

train_data = df1[
    df1["Year_i"].isin(training_years)
].copy()

test_data = df1[
    df1["Year_i"] == test_year
].copy()

print("\nTraining years:")
print(training_years)

print("\nTraining rows:", len(train_data))
print("2024 test rows:", len(test_data))
```

```python
# ============================================================
# 8. CHECK COVID YEARS
# ============================================================

covid_data = df1[
    df1["Year_i"].isin([2020, 2021])
]

print("\nCOVID rows excluded from training:", len(covid_data))
```

```python
# ============================================================
# 9. PREDICT 2024
# ============================================================

predictions = []

# Group the training data by the exact CPI series.
#
# January 2014
# January 2015
# ...
# January 2023
# are used to predict January 2024.


grouped_train = train_data.groupby(keys, sort=False)

total_series = len(grouped_train)
processed_series = 0

print("\nStarting 2024 prediction...")
print("Total historical series:", total_series)

for series_key, train_group in grouped_train:

    processed_series += 1

    # --------------------------------------------------------
    # Get corresponding 2024 observation
    # --------------------------------------------------------

    condition = np.ones(len(test_data), dtype=bool)

    for col, value in zip(keys, series_key):
        condition &= (
            test_data[col].values == value
        )

    actual_2024 = test_data.loc[condition]

    # If this series does not exist in 2024,
    # there is nothing to predict.
    if len(actual_2024) == 0:
        continue

    # --------------------------------------------------------
    # Remove missing CPI values
    # --------------------------------------------------------

    train_group = train_group.dropna(
        subset=["Index_i"]
    )

    # We need at least 2 historical observations
    # for linear regression.
    if len(train_group) < 2:
        continue

    # --------------------------------------------------------
    # Prepare X and y
    # --------------------------------------------------------

    X = train_group[["Year_i"]]
    y = train_group["Index_i"]

    # --------------------------------------------------------
    # Linear Regression
    # --------------------------------------------------------

    model = LinearRegression()

    model.fit(X, y)

    # --------------------------------------------------------
    # Predict 2024
    # --------------------------------------------------------

    predicted_2024 = model.predict(
        np.array([[2024]])
    )[0]

    # --------------------------------------------------------
    # Store result
    # --------------------------------------------------------

    for _, actual_row in actual_2024.iterrows():

        result = actual_row.to_dict()

        result["Predicted_Index_2024"] = predicted_2024

        # Number of historical observations used
        result["Training_Observations"] = len(train_group)

        # Model slope
        result["Trend_Slope"] = model.coef_[0]

        predictions.append(result)

    # Progress
    if processed_series % 1000 == 0:
        print(
            f"Processed {processed_series}/{total_series} series"
        )
```

```python
# ============================================================
# 10. CREATE PREDICTION DATAFRAME
# ============================================================

predicted_2024 = pd.DataFrame(predictions)

print("\nPrediction completed!")

print(
    "Predicted 2024 rows:",
    len(predicted_2024)
)
```

```python
# ============================================================
# 11. CHECK PREDICTIONS
# ============================================================

print("\nFirst 10 predictions:")

print(
    predicted_2024[
        keys
        + [
            "Index_i",
            "Predicted_Index_2024",
            "Training_Observations",
            "Trend_Slope"
        ]
    ].head(10)
)

```

```python
# ============================================================
# 12. CALCULATE PREDICTION ERROR
# ============================================================

actual = predicted_2024["Index_i"]

predicted = predicted_2024[
    "Predicted_Index_2024"
]

# Error
predicted_2024["Prediction_Error"] = (
    actual - predicted
)

# Absolute error
predicted_2024["Absolute_Error"] = (
    abs(actual - predicted)
)

# Percentage error
predicted_2024["Percentage_Error"] = (
    abs(actual - predicted) / actual
) * 100

# ============================================================
# 13. EVALUATION METRICS
# ============================================================

MAE = mean_absolute_error(
    actual,
    predicted
)

MSE = mean_squared_error(
    actual,
    predicted
)

RMSE = np.sqrt(MSE)

MAPE = np.mean(
    np.abs(
        (actual - predicted) / actual
    )
) * 100

R2 = r2_score(
    actual,
    predicted
)

# Approximate accuracy
accuracy = 100 - MAPE

# ============================================================
# 14. DISPLAY METRICS
# ============================================================

print("\n" + "=" * 60)
print("2024 CPI PREDICTION EVALUATION")
print("=" * 60)

print(f"MAE  : {MAE:.4f}")
print(f"MSE  : {MSE:.4f}")
print(f"RMSE : {RMSE:.4f}")
print(f"MAPE : {MAPE:.4f}%")
print(f"R²   : {R2:.4f}")

print(
    f"Approximate Accuracy : {accuracy:.4f}%"
)

print("=" * 60)

```

```python
# ============================================================
# 15. CREATE METRICS DATAFRAME
# ============================================================

metrics = pd.DataFrame({
    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "MAPE",
        "R2",
        "Approximate Accuracy"
    ],

    "Value": [
        MAE,
        MSE,
        RMSE,
        MAPE,
        R2,
        accuracy
    ]
})

print("\nEvaluation Metrics:")
print(metrics)

```

# Now Prediction for year 2025 using the same accuracy that have been for 2024

```python
import numpy as np
import pandas as pd
from lightgbm import LGBMRegressor

# ============================================================
# 1. ESSENTIAL PREPARATION & SORTING
# ============================================================
series_keys = ["State_i", "Sector", "Group_i", "SubGroup"]
for col in series_keys:
    df1[col] = df1[col].astype(str).str.strip()

month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
month_map = {m: float(i) for i, m in enumerate(month_order, 1)}
df1["Month_num"] = df1["Month_i"].map(month_map).astype(float)
df1["Month_i"] = pd.Categorical(df1["Month_i"], categories=month_order, ordered=True)

df1["Year_i"] = pd.to_numeric(df1["Year_i"], errors="coerce").astype(int)
df1["Inflation"] = pd.to_numeric(df1["Inflation"], errors="coerce").astype(float)
df1["Index_i"] = pd.to_numeric(df1["Index_i"], errors="coerce").astype(float)

# Chronological sort required for accurate lags
df1.sort_values(by=series_keys + ["Year_i", "Month_num"], inplace=True)
df1.reset_index(drop=True, inplace=True)

# ============================================================
# 2. COMPUTE MISSING FEATURE COLUMNS & TARGET
# ============================================================
median_index_2024 = df1.loc[df1["Year_i"] == 2024, "Index_i"].median()

df1["sin_month"] = np.sin(2 * np.pi * df1["Month_num"] / 12)
df1["cos_month"] = np.cos(2 * np.pi * df1["Month_num"] / 12)

df1["Inflation_Lag_12M"] = df1.groupby(series_keys + ["Month_num"])["Inflation"].shift(1)
df1["Inflation_Lag_24M"] = df1.groupby(series_keys + ["Month_num"])["Inflation"].shift(2)
df1["Index_Lag_12M"] = df1.groupby(series_keys + ["Month_num"])["Index_i"].shift(1).fillna(median_index_2024)
df1["Log_Index_Lag_12M"] = np.log1p(df1["Index_Lag_12M"])

macro_mean = df1.groupby(["SubGroup", "Year_i", "Month_num"])["Inflation"].transform("mean")
df1["Subgroup_Macro_Mean_Lag_12M"] = df1.groupby(series_keys + ["Month_num"])[macro_mean.name].shift(1)

p01 = df1["Inflation"].quantile(0.01)
p99 = df1["Inflation"].quantile(0.99)
df1["Inflation_Target"] = df1["Inflation"].clip(p01, p99)

# ============================================================
# 3. DEFINE FEATURE LIST & CATEGORIES
# ============================================================
cat_cols = ["Month_i", "State_i", "Sector", "Group_i", "SubGroup"]
for col in cat_cols:
    df1[col] = df1[col].astype("category")

feature_cols = cat_cols + [
    "sin_month",
    "cos_month",
    "Inflation_Lag_12M",
    "Inflation_Lag_24M",
    "Log_Index_Lag_12M",
    "Subgroup_Macro_Mean_Lag_12M",
]

# ============================================================
# 4. FIT ON FULL HISTORY (2016-2024, EXCLUDING 2020 & 2021)
# ============================================================
full_train_mask = (
    (df1["Year_i"] >= 2016)
    & (df1["Year_i"] <= 2024)
    & (~df1["Year_i"].isin([2020, 2021]))
    & df1[feature_cols + ["Inflation_Target"]].notna().all(axis=1)
)

final_model = LGBMRegressor(
    objective="huber",
    n_estimators=350,
    learning_rate=0.03,
    num_leaves=31,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    verbose=-1,
)
final_model.fit(
    df1.loc[full_train_mask, feature_cols],
    df1.loc[full_train_mask, "Inflation_Target"],
)

# ============================================================
# 5. PREDICT & MERGE 2025 IN-PLACE
# ============================================================
idx_2025 = df1.index[df1["Year_i"] == 2025]

df1.loc[idx_2025, "Inflation"] = final_model.predict(
    df1.loc[idx_2025, feature_cols]
)
df1.loc[idx_2025, "Index_i"] = df1.loc[idx_2025, "Index_Lag_12M"] * (
    1 + df1.loc[idx_2025, "Inflation"] / 100
)
df1["Index_i"] = df1["Index_i"].fillna(median_index_2024)

# ============================================================
# 6. RESTORE STANDARD COLUMNS & SORT (YEAR + JAN TO DEC)
# ============================================================
final_columns = [
    "BYear", "Year_i", "Month_i", "State_i", "Sector",
    "Group_i", "SubGroup", "Index_i", "Inflation",
]

df1 = df1[final_columns].sort_values(
    by=["Year_i", "Month_i", "State_i", "Sector", "SubGroup"]
).reset_index(drop=True)

# ============================================================
# 7. DISPLAY 15 POINTERS OF 2025 FORECASTED DATA
# ============================================================
df1[df1["Year_i"] == 2025].head(50)
```
